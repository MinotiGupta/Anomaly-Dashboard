```python
import os
import math
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from dataclasses import dataclass
from enum import Enum
from typing import Optional, List, Tuple

# =============================================================================
# 1. ENUMS AND CONFIGURATION 
# =============================================================================
class FlagClass(Enum):
    VALID = "VALID"
    IMPLAUSIBLE = "IMPLAUSIBLE"

@dataclass
class ChannelConfig:
    tau: float                  
    tau_uncertainty_margin: float 
    T_sink: float               
    T_drive_max: float          
    sensor_min: float           
    sensor_max: float           

# =============================================================================
# 2. PHYSICS HELPER FUNCTIONS
# =============================================================================
def calc_slope_bounds(T: float, T_sink: float, T_drive_max: float, tau: float) -> Tuple[float, float]:
    """Rate of change boundaries (dT/dt) based on sensor thermal mass (K/s)."""
    max_slope = (T_drive_max - T) / tau
    min_slope = -(T - T_sink) / tau
    return min_slope, max_slope

# =============================================================================
# 3. STREAMING PHYSICS ENGINE (Deterministic Veto)
# =============================================================================
class TemperatureChannel:
    def __init__(self, config: ChannelConfig):
        self.config = config
        self.t_prev: Optional[float] = None
        self.T_prev: Optional[float] = None

    def process(self, t_now: float, T_raw: float):
        flags = []
        is_vetoed = False

        if T_raw < self.config.sensor_min or T_raw > self.config.sensor_max:
            is_vetoed, flags = True, flags + ["Hard sensor limit violated"]

        if T_raw < self.config.T_sink or T_raw > self.config.T_drive_max:
            is_vetoed, flags = True, flags + ["Thermodynamic bound violated"]

        dt = (t_now - self.t_prev) if self.t_prev is not None else None
        
        if self.T_prev is not None and dt is not None and dt > 0:
            dT_dt = (T_raw - self.T_prev) / dt
            min_slope, max_slope = calc_slope_bounds(
                self.T_prev, self.config.T_sink, self.config.T_drive_max, self.config.tau
            )
            if dT_dt > (max_slope * self.config.tau_uncertainty_margin) or dT_dt < (min_slope * self.config.tau_uncertainty_margin):
                is_vetoed, flags = True, flags + [f"Physics Rate Violation ({dT_dt:.2f} K/s)"]

        flag_class = FlagClass.IMPLAUSIBLE if is_vetoed else FlagClass.VALID

        # Vetoed data does not update state
        if not is_vetoed:
            self.t_prev, self.T_prev = t_now, T_raw

        return flag_class.value, " | ".join(flags)

# =============================================================================
# 4. HYBRID BATCH PROCESSING (Physics + Residual Learning)
# =============================================================================
def process_and_plot_file(file_path: str, config: ChannelConfig):
    if not os.path.exists(file_path):
        print(f"Error: File not found -> {os.path.basename(file_path)}")
        return

    # 1. Parse Data
    header_idx = 0
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        for idx, line in enumerate(f):
            if any(k in line.lower() for k in ['scan swee', 'scan number', '101time', 'start time']):
                header_idx = idx
                break
                
    df = pd.read_csv(file_path, skiprows=header_idx)
    df.columns = df.columns.str.strip()
    df = df.dropna(how='all', axis=1)
    
    # Identify Time Column
    time_cols = [col for col in df.columns if 'time' in col.lower()]
    time_col = time_cols[0] if time_cols else df.columns[0]
    
    # Identify Sensor Column (with Robust Fallback)
    sensor_cols = [col for col in df.columns if any(k in col.lower() for k in ['°c', '101', 'ch_00', 'temp', '(c)'])]
    if sensor_cols:
        sensor_col = sensor_cols[-1]
    else:
        possible_sensors = [col for col in df.columns if col != time_col]
        sensor_col = possible_sensors[-1] if possible_sensors else df.columns[-1]
        
    df[sensor_col] = pd.to_numeric(df[sensor_col], errors='coerce')
    
    # Check if there's actually any data to parse
    time_data = df[time_col].dropna()
    if time_data.empty:
        print(f"❌ Error: Time column '{time_col}' is completely empty in {os.path.basename(file_path)}")
        return

    # Handle Time Formats
    first_time_val = str(time_data.iloc[0])
    if ' ' in first_time_val or '-' in first_time_val:
        df['Timestamp'] = pd.to_datetime(df[time_col], errors='coerce')
    elif ':' in first_time_val:
        df['Timestamp'] = pd.to_timedelta("00:" + df[time_col].astype(str), errors='coerce')
    else:
        # Fallback if the time is already formatted as raw seconds
        df['Seconds_Elapsed'] = pd.to_numeric(df[time_col], errors='coerce')
        df['Timestamp'] = pd.to_datetime(df['Seconds_Elapsed'], unit='s', errors='coerce')

    # Drop invalid rows
    if 'Seconds_Elapsed' in df.columns:
        df_clean = df.dropna(subset=['Seconds_Elapsed', sensor_col]).copy()
    else:
        df_clean = df.dropna(subset=['Timestamp', sensor_col]).copy()

    # THE FIX: Guard against completely empty dataframes after dropping NaNs
    if df_clean.empty:
        print(f"❌ Error: No valid data remains after parsing timestamps in {os.path.basename(file_path)}")
        return

    # Calculate continuous elapsed seconds if not already done
    if 'Seconds_Elapsed' not in df_clean.columns:
        if pd.api.types.is_timedelta64_dtype(df_clean['Timestamp']):
            df_clean['Seconds_Elapsed'] = df_clean['Timestamp'].dt.total_seconds()
        else:
            df_clean['Seconds_Elapsed'] = (df_clean['Timestamp'] - df_clean['Timestamp'].iloc[0]).dt.total_seconds()

    # 2. RUN PHYSICS ENGINE
    engine = TemperatureChannel(config)
    physics_results = [engine.process(row['Seconds_Elapsed'], row[sensor_col]) for _, row in df_clean.iterrows()]
    df_clean['Physics_Flag'], df_clean['Physics_Reason'] = zip(*physics_results)

    # 3. RUN RESIDUAL LEARNING (Dynamic Baseline & MAD Thresholding)
    df_clean['Valid_Sensor_Data'] = np.where(df_clean['Physics_Flag'] == 'VALID', df_clean[sensor_col], np.nan)
    df_clean['Valid_Sensor_Data'] = df_clean['Valid_Sensor_Data'].ffill().bfill()

    span_scans = max(1, int(config.tau)) 
    df_clean['Learned_Baseline'] = df_clean['Valid_Sensor_Data'].ewm(span=span_scans, adjust=False).mean()
    
    df_clean['Residual'] = (df_clean[sensor_col] - df_clean['Learned_Baseline']).abs()
    
    rolling_window = 20
    df_clean['Rolling_MAD'] = df_clean['Residual'].rolling(window=rolling_window, min_periods=1).apply(
        lambda x: np.median(np.abs(x - np.median(x)))
    )
    
    noise_multiplier = 4.0
    minimum_noise_floor = 0.5 
    df_clean['Dynamic_Threshold'] = np.maximum(df_clean['Rolling_MAD'] * 1.4826 * noise_multiplier, minimum_noise_floor)
    
    df_clean['Residual_Anomaly'] = np.where(df_clean['Residual'] > df_clean['Dynamic_Threshold'], -1, 1)

    # 4. SUMMARY METRICS
    phys_anomalies = df_clean[df_clean['Physics_Flag'] == 'IMPLAUSIBLE']
    res_anomalies = df_clean[(df_clean['Residual_Anomaly'] == -1) & (df_clean['Physics_Flag'] == 'VALID')]
    
    print(f"\n{'='*60}")
    print(f"FILE: {os.path.basename(file_path)}")
    print(f"{'-'*60}")
    print(f"Total Data Points       : {len(df_clean)}")
    print(f"Physics Engine Vetoes   : {len(phys_anomalies)}")
    print(f"Residual Anomalies      : {len(res_anomalies)}")
    print(f"{'='*60}\n")
    
    # 5. GENERATE PLOT
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=df_clean['Timestamp'], y=df_clean['Learned_Baseline'], 
        mode='lines', name='Learned Baseline (EMA)',
        line=dict(color='rgba(200, 200, 200, 0.8)', width=3), hoverinfo='skip'
    ))
    
    valid_data = df_clean[(df_clean['Physics_Flag'] == 'VALID') & (df_clean['Residual_Anomaly'] == 1)]
    fig.add_trace(go.Scatter(
        x=valid_data['Timestamp'], y=valid_data[sensor_col], 
        mode='lines', name='Normal Signal',
        line=dict(color='#1f77b4', width=1.5)
    ))

    if not res_anomalies.empty:
        fig.add_trace(go.Scatter(
            x=res_anomalies['Timestamp'], y=res_anomalies[sensor_col], 
            mode='markers', name='Residual Anomaly',
            marker=dict(color='purple', size=6, symbol='square', line=dict(width=1, color='darkviolet')),
            hovertemplate="Time: %{x}<br>Temp: %{y}°C<br>Flag: Deviation exceeded dynamic threshold<extra></extra>"
        ))
    
    if not phys_anomalies.empty:
        fig.add_trace(go.Scatter(
            x=phys_anomalies['Timestamp'], y=phys_anomalies[sensor_col], 
            mode='markers', name='Physics Veto (Impossible)',
            marker=dict(color='red', size=8, symbol='x', line=dict(width=1.5, color='darkred')),
            text=phys_anomalies['Physics_Reason'],
            hovertemplate="Time: %{x}<br>Temp: %{y}°C<br>Reason: %{text}<extra></extra>"
        ))
        
    fig.update_layout(
        title=f"Hybrid Anomaly Detection: {os.path.basename(file_path)}",
        xaxis_title="Time", yaxis_title="Temperature (°C)",
        template='plotly_white', hovermode='x unified',
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01),
        height=500, width=1200
    )
    
    if pd.api.types.is_timedelta64_dtype(df_clean['Timestamp']):
        fig.update_xaxes(tickformat="%M:%S")
        
    fig.show()

# =============================================================================
# 5. EXECUTION LOOP
# =============================================================================
file_paths = [
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\2 2025-11-21 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\03-11-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\03-11-2025 2 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\06-11-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\07-11-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\13-11-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\14-11-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\15-09-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\16-09-2025 0 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\17-11-2025.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\18-09-2025 0.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\21-11-2025 0 2025-11-21.csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\2025-11-15 .csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\2025-11-17 .csv",
    r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\2025-11-19  0.csv"
]

shared_config = ChannelConfig(
    tau=15.0, 
    tau_uncertainty_margin=1.2, 
    T_sink=20.0, 
    T_drive_max=2200.0, 
    sensor_min=-50.0, 
    sensor_max=1370.0
)

for fp in file_paths:
    process_and_plot_file(fp, shared_config)
```
