```python
"""
Ai4Energy Anomaly Detection Module
Implementation based strictly on physical priors and hardware constraints[cite: 26, 28].
No unsupervised machine learning or history-dependent baseline models are used[cite: 28, 43].
"""

import math
import numpy as np
from dataclasses import dataclass, field
from enum import Enum
from typing import Optional, List, Dict, Any, Tuple
import pytest

# =============================================================================
# 1. ENUMS AND CONFIGURATION (Deliverable 1)
# =============================================================================

class FlagClass(Enum):
    VALID = "VALID"
    UNUSUAL = "UNUSUAL"         # Physical but statistically surprising[cite: 29]
    SUSPICIOUS = "SUSPICIOUS"   # Consistent with degradation/instrument signature[cite: 29]
    IMPLAUSIBLE = "IMPLAUSIBLE" # Physics or datasheet constraint violation (Veto)[cite: 29]

@dataclass
class FlagResult:
    raw_value: float            # Must never be filtered[cite: 30]
    flag: FlagClass
    reasons: List[str]
    residual: float
    threshold: float
    config_version: str

@dataclass
class ChannelConfig:
    version: str
    sensor_type: str            # 'K_TYPE_MAX31855', 'K_TYPE_MAX31856', 'DS18B20', 'SHT45'
    tau: float                  # Thermal time constant (seconds)[cite: 34]
    tau_uncertainty_margin: float # Multiplier for tau variance
    T_sink: float               # Coldest available sink (K or °C)[cite: 40]
    T_drive_max: float          # Adiabatic flame temp / max source (K or °C)[cite: 40]
    sensor_min: float           # e.g., -55 for DS18B20[cite: 40]
    sensor_max: float           # e.g., 1370 for K-type[cite: 40]
    resolution: float           # LSB in °C[cite: 31, 38]
    base_noise_floor: float     # Datasheet or pre-run measured electrical noise[cite: 40, 42]
    nominal_dt: float           # Expected sample interval (seconds)
    twin_channel_tau: Optional[float] = None
    is_ds18b20: bool = False    # To account for 1-sample conversion delay[cite: 33]
    # ASSUMPTION TO VERIFY: T_drive_max must be supplied by the combustion team per rig.
    # ASSUMPTION TO VERIFY: IEC 60584 tolerances for the specific wire class.

# =============================================================================
# 2. INTEGRITY DECODERS (Deliverable 3)
# =============================================================================

def check_max31855(spi_word: int) -> Tuple[bool, List[str]]:
    """
    Layer 0 Integrity for MAX31855. No CRC exists[cite: 31].
    ASSUMPTION TO VERIFY: Exact reserved bit positions against current datasheet[cite: 31].
    """
    faults = []
    if spi_word == 0x00000000 or spi_word == 0xFFFFFFFF:
        faults.append("All zeros/ones outright rejection") #[cite: 31]
        return False, faults
    
    # Mocking bit positions (assumes standard 32-bit MAX31855 word)
    reserved_bit_1 = (spi_word >> 17) & 0x1
    reserved_bit_2 = (spi_word >> 31) & 0x1 
    if reserved_bit_1 or reserved_bit_2:
        faults.append("Reserved bits set (corrupt transfer)") #[cite: 31]

    fault_bit = (spi_word >> 16) & 0x1
    scv = (spi_word >> 2) & 0x1  # Short to VCC
    scg = (spi_word >> 1) & 0x1  # Short to GND
    oc = spi_word & 0x1          # Open Circuit
    
    if fault_bit != (scv | scg | oc):
        faults.append("Summary fault bit disagrees with cause bits")  #[cite: 31]
        
    return len(faults) == 0, faults

def check_max31856(fault_register: int) -> Tuple[bool, List[str]]:
    """
    Layer 0 Integrity for MAX31856. Decodes 8 distinct fault conditions[cite: 32].
    """
    faults = []
    if fault_register & 0x01: faults.append("Open Circuit") #[cite: 32]
    if fault_register & 0x02: faults.append("Overvoltage / Undervoltage") #[cite: 32]
    if fault_register & 0x04: faults.append("Thermocouple Low") #[cite: 32]
    if fault_register & 0x08: faults.append("Thermocouple High") #[cite: 32]
    if fault_register & 0x10: faults.append("Cold Junction Low") #[cite: 32]
    if fault_register & 0x20: faults.append("Cold Junction High") #[cite: 32]
    if fault_register & 0x40: faults.append("TCRange")
    if fault_register & 0x80: faults.append("CJRange")
    return len(faults) == 0, faults

def check_ds18b20(scratchpad: bytes, current_temp: float, conversions_done: int) -> Tuple[bool, List[str]]:
    """
    Layer 0 Integrity for DS18B20 1-Wire bus[cite: 32].
    """
    faults = []
    # Note: proper CRC-8 calculation (poly 0x8C) would iterate over the 8 bytes.
    if len(scratchpad) == 9:
        if scratchpad == b'\x00'*9 or scratchpad == b'\xff'*9:
            faults.append("Bus failure (all zeros/ones)") #[cite: 32]
    
    if current_temp == 85.0 and conversions_done == 0:
        faults.append("Power-on value (85.0 C) with no conversion completed") #[cite: 32]
        
    return len(faults) == 0, faults

# =============================================================================
# 3. PHYSICS HELPER FUNCTIONS (Deliverable 4)
# =============================================================================

def calc_alpha(dt: float, tau: float) -> float:
    """Discrete exponential filter coefficient. Section 4.3[cite: 41]."""
    return math.exp(-dt / tau)

def calc_slope_bounds(T: float, T_sink: float, T_drive_max: float, tau: float) -> Tuple[float, float]:
    """
    Rate of change boundaries (dT/dt) based on sensor thermal mass. Section 3.2, 4.2[cite: 35, 40].
    Returns (min_slope, max_slope) in K/s.
    """
    max_slope = (T_drive_max - T) / tau
    min_slope = -(T - T_sink) / tau
    return min_slope, max_slope

def calc_radiation_error(T_tc: float, T_w: float, h: float, epsilon: float = 0.6) -> float:
    """
    Radiation bias (K). Diagnostic only. Section 3.3[cite: 36].
    """
    sigma = 5.67e-8
    return (epsilon * sigma * (T_tc**4 - T_w**4)) / h

def calc_conduction_error(T_mount: float, T_g: float, h: float, P: float, k: float, Ac: float, L: float) -> float:
    """
    Conduction (stem) error (K). Section 3.3[cite: 37].
    """
    m = math.sqrt((h * P) / (k * Ac))
    return (T_mount - T_g) / math.cosh(m * L)

def calc_recovery_effect(T_static: float, u: float, cp: float, r: float = 0.75) -> float:
    """
    Velocity recovery effect (K). Relevant > 50 m/s. Section 3.3[cite: 37].
    """
    return T_static + (r * u**2) / (2 * cp)

def calc_biot_number(h: float, L_c: float, k_s: float) -> float:
    """
    Biot number to validate lumped capacitance model. Warn if > 0.1. Section 3.1[cite: 34].
    """
    return (h * L_c) / k_s

def calc_quantisation_noise(resolution: float) -> float:
    """
    Lower bound for noise scale based on quantization. Section 4.6[cite: 42].
    """
    return resolution / math.sqrt(12)

def calc_daq_bandwidth(nplc: int, f_mains: float = 50.0) -> float:
    """
    Effective bandwidth (Hz) of DAQ970A moving-average integration. Section 5.1[cite: 44].
    """
    t_int = nplc / f_mains
    return 0.44 / t_int

def fit_tau(times: np.ndarray, temps: np.ndarray, T_final: float) -> float:
    """
    In-situ estimation of tau from a transient response. Section 4.4[cite: 41].
    Skips first 10% to avoid two-lump sheath dynamics[cite: 35].
    """
    idx_10pct = len(times) // 10
    t_fit = times[idx_10pct:]
    T_diff = np.abs(temps[idx_10pct:] - T_final)
    # Mask zeros to avoid log(0)
    valid = T_diff > 1e-6
    if np.sum(valid) < 2:
        return 0.0
    
    # Linear fit of ln(dT) vs time -> slope is -1/tau
    slope, _ = np.polyfit(t_fit[valid], np.log(T_diff[valid]), 1)
    return -1.0 / slope if slope < 0 else float('inf')

# =============================================================================
# 4. STREAMING ENGINE (Deliverable 2)
# =============================================================================

class TemperatureChannel:
    def __init__(self, config: ChannelConfig):
        self.config = config
        
        # State tracking
        self.t_prev: Optional[float] = None
        self.T_prev: Optional[float] = None
        self.twin_y_hat_prev: Optional[float] = None
        
        # Robust statistics buffers (Quiescent windows)
        self.quiescent_buffer: List[float] = []
        self.current_mad_scale: float = self.config.base_noise_floor
        
        # CUSUM State
        self.cusum_S_pos = 0.0
        self.cusum_S_neg = 0.0

        # DS18B20 delay tracking
        self.ds18b20_prev_T: Optional[float] = None
        
    def _update_mad_scale(self):
        """Update noise scale using Median Absolute Deviation[cite: 42]."""
        if len(self.quiescent_buffer) > 10:
            med = np.median(self.quiescent_buffer)
            mad = np.median(np.abs(np.array(self.quiescent_buffer) - med))
            # Fallback to datasheet floor if MAD is zero
            self.current_mad_scale = max(self.config.base_noise_floor, 1.4826 * mad)

    def process(self, 
                t_now: float, 
                T_raw: float, 
                integrity_ok: bool, 
                integrity_faults: List[str],
                twin_T: Optional[float] = None) -> FlagResult:
        
        flags = []
        is_vetoed = False
        residual = 0.0
        threshold = 0.0

        # 1. Structural / Integrity Checks (Layer 0)[cite: 30]
        if not integrity_ok:
            is_vetoed = True
            flags.extend(integrity_faults)
            
        # 2. Thermodynamic Bounds[cite: 40]
        if T_raw < self.config.sensor_min or T_raw > self.config.sensor_max:
            is_vetoed = True
            flags.append(f"Hard sensor limit violated: {T_raw}") #[cite: 40]
            
        if T_raw < self.config.T_sink or T_raw > self.config.T_drive_max:
            is_vetoed = True
            flags.append(f"Thermodynamic bound violated: {T_raw}") #[cite: 40]

        # 3. Dynamical Physics (Rate of Change)[cite: 40]
        dt = (t_now - self.t_prev) if self.t_prev is not None else None
        
        if self.T_prev is not None and dt is not None and dt > 0:
            dT_dt = (T_raw - self.T_prev) / dt
            min_slope, max_slope = calc_slope_bounds()
```

```python
import os
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from typing import Optional, Tuple, List

# ====================================================================
# 1. ROBUST DATA LOADER
# ====================================================================
def load_and_parse_csv(file_path: str) -> Tuple[pd.DataFrame, str, str]:
    """
    Dynamically locates the table header, loads the CSV,
    and isolates the time and sensor columns using a robust fallback.
    """
    header_row_index = 0
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        for idx, line in enumerate(f):
            if any(k in line.lower() for k in ['scan num', '101 (', 'scan swee', 'timestamp', 'start time']):
                header_row_index = idx
                break
                
    df = pd.read_csv(file_path, skiprows=header_row_index)
    df.columns = df.columns.str.strip()
    df = df.dropna(how='all', axis=1).dropna(how='all', axis=0)
    
    # 1. Identify the time column first
    time_cols = [col for col in df.columns if any(k in col.lower() for k in ['time', 'swee', 'timestamp', 'start'])]
    time_col = time_cols[0] if time_cols else df.columns[0]
    
    # 2. Identify the sensor column
    sensor_cols = [col for col in df.columns if any(k in col.lower() for k in ['°c', '101', 'ch_00', 'temp', '(c)'])]
    
    # Fallback: If strict matching fails (due to encoding or renaming), 
    # exclude time/scan columns and grab the first available data column.
    if not sensor_cols:
        exclude = time_cols + [col for col in df.columns if 'scan' in col.lower() or 'num' in col.lower() or 'unnamed' in col.lower()]
        possible_sensors = [col for col in df.columns if col not in exclude]
        
        if possible_sensors:
            sensor_cols = [possible_sensors[0]]
        else:
            # If it still fails, print the actual columns to the terminal so we can see what's happening
            raise ValueError(f"Could not detect a valid temperature channel header. Columns found: {list(df.columns)}")
            
    sensor_col = sensor_cols[0]
    df[sensor_col] = pd.to_numeric(df[sensor_col], errors='coerce')
    
    # 3. Create a monotonic time index in seconds for accurate dt calculation
    try:
        # Attempt to parse mm:ss.s or standard timestamps
        if df[time_col].dtype == object and ':' in str(df[time_col].iloc[0]):
            df['Seconds_Elapsed'] = pd.to_timedelta("00:" + df[time_col].astype(str)).dt.total_seconds()
        else:
            df['Seconds_Elapsed'] = pd.to_numeric(df[time_col], errors='coerce')
    except Exception:
        # Fallback to row index if time parsing fails
        df['Seconds_Elapsed'] = np.arange(len(df), dtype=float)
        
    return df, sensor_col, time_col

# ====================================================================
# 2. STATE-AWARE PHYSICS ENGINE
# ====================================================================
def apply_physics_model(df: pd.DataFrame, sensor_col: str) -> pd.DataFrame:
    """
    Evaluates each sample against thermodynamic amplitude and rate limits[cite: 40].
    Strictly annotates data without filtering[cite: 29]. Vetoed samples do not update baseline state[cite: 42].
    """
    # Physics Configuration
    TAU = 15.0               # Thermal time constant for sheathed probe[cite: 34]
    TAU_MARGIN = 1.2         # Margin for tau uncertainty
    T_DRIVE_MAX = 2200.0     # Adiabatic flame temperature[cite: 40]
    T_SINK = 20.0            # Coldest available sink[cite: 40]
    
    processed_df = df.copy()
    processed_df['Quality_Flag'] = 'VALID'
    processed_df['dT_dt'] = np.nan
    
    # State tracking variables
    last_valid_T: Optional[float] = None
    last_valid_t: Optional[float] = None
    
    for idx, row in processed_df.iterrows():
        T_raw = row[sensor_col]
        t_now = row['Seconds_Elapsed']
        
        if pd.isna(T_raw):
            continue
            
        is_vetoed = False
        flags: List[str] = []
        
        # 1. Absolute Thermodynamic Amplitude Bounds[cite: 40]
        if T_raw > T_DRIVE_MAX or T_raw < -50.0:
            is_vetoed = True
            flags.append("Amplitude Out of Bounds")
            
        # 2. First-Order Lag Rate Limits[cite: 40]
        if not is_vetoed and last_valid_T is not None and last_valid_t is not None:
            dt = t_now - last_valid_t
            if dt > 0:
                dT_dt = (T_raw - last_valid_T) / dt
                processed_df.at[idx, 'dT_dt'] = dT_dt
                
                # Calculate absolute physics speed limits for this specific temperature
                max_slope = ((T_DRIVE_MAX - last_valid_T) / TAU) * TAU_MARGIN
                min_slope = (-(last_valid_T - T_SINK) / TAU) * TAU_MARGIN
                
                if dT_dt > max_slope or dT_dt < min_slope:
                    is_vetoed = True
                    flags.append(f"Rate Violation ({dT_dt:.1f} K/s)")

        # 3. State Update & Annotation[cite: 29, 42]
        if is_vetoed:
            processed_df.at[idx, 'Quality_Flag'] = "IMPLAUSIBLE: " + " | ".join(flags)
            # Architectural Rule: Vetoed data does not update state[cite: 42]
        else:
            last_valid_T = T_raw
            last_valid_t = t_now
            
    return processed_df

# ====================================================================
# 3. VISUALIZATION ENGINE
# ====================================================================
def plot_physics_results(df: pd.DataFrame, sensor_col: str, time_col: str, file_name: str):
    """Generates an interactive Plotly overlay of valid vs implausible data."""
    anomalies = df[df['Quality_Flag'] != 'VALID']
    valid_data = df[df['Quality_Flag'] == 'VALID']
    
    fig = go.Figure()
    
    # 1. Base Signal Line (Connects all points lightly to show full raw trajectory)
    fig.add_trace(go.Scatter(
        x=df[time_col], 
        y=df[sensor_col], 
        mode='lines', 
        name='Raw Trajectory', 
        line=dict(color='lightgrey', width=1.5),
        hoverinfo='skip'
    ))
    
    # 2. Valid Physical Data
    fig.add_trace(go.Scatter(
        x=valid_data[time_col], 
        y=valid_data[sensor_col], 
        mode='markers', 
        name='Valid (Obeys Physics)', 
        marker=dict(color='#1f77b4', size=4)
    ))
    
    # 3. Implausible Anomalies (Vetoed)
    if not anomalies.empty:
        fig.add_trace(go.Scatter(
            x=anomalies[time_col], 
            y=anomalies[sensor_col], 
            mode='markers', 
            name='IMPLAUSIBLE (Physics Violation)', 
            marker=dict(color='red', size=8, symbol='x', line=dict(width=1.5, color='darkred')),
            text=anomalies['Quality_Flag'],
            hovertemplate="Time: %{x}<br>Temp: %{y}°C<br>Flag: %{text}<extra></extra>"
        ))
        
    fig.update_layout(
        title=f"Physics-Based Anomaly Validation: {file_name}",
        xaxis_title="Time / Scan",
        yaxis_title="Temperature (°C)",
        template='plotly_white',
        hovermode="x unified",
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
    )
    
    fig.show()

# ====================================================================
# 4. EXECUTION
# ====================================================================
file_path = r"C:\Users\hari7\Documents\Anamoly Detection\OneDrive_1_10-6-2026\03-11-2025 2 0.csv"

if os.path.exists(file_path):
    print(f"Loading {os.path.basename(file_path)}...")
    
    # 1. Extract
    df_raw, target_sensor, time_col = load_and_parse_csv(file_path)
    print(f"Successfully loaded. Processing {len(df_raw)} samples on channel '{target_sensor}'.")
    
    # 2. Process
    df_validated = apply_physics_model(df_raw, target_sensor)
    
    anomalies_count = (df_validated['Quality_Flag'] != 'VALID').sum()
    print(f"Physics Engine complete. Flagged {anomalies_count} implausible anomalies.")
    
    # 3. Visualize
    plot_physics_results(df_validated, target_sensor, time_col, os.path.basename(file_path))
else:
    print(f"File not found: {file_path}")
```

```python
import os
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from typing import Optional, Tuple, List

# ====================================================================
# 1. ROBUST DATA LOADER
# ====================================================================
def load_and_parse_csv(file_path: str) -> Tuple[pd.DataFrame, str, str]:
    """
    Dynamically locates the table header, loads the CSV,
    and isolates the scan and sensor columns.
    """
    header_row_index = 0
    with open(file_path, 'r', encoding='utf-8', errors='ignore') as f:
        for idx, line in enumerate(f):
            if any(k in line.lower() for k in ['scan num', '101 (', 'scan swee', 'timestamp', 'start time']):
                header_row_index = idx
                break
                
    df = pd.read_csv(file_path, skiprows=header_row_index)
    df.columns = df.columns.str.strip()
    df = df.dropna(how='all', axis=1).dropna(how='all', axis=0)
    
    # 1. Identify the Scan Number column for a guaranteed clean X-axis
    scan_cols = [col for col in df.columns if 'scan num' in col.lower()]
    scan_col = scan_cols[0] if scan_cols else df.columns[0]
    
    # Force the scan column to numeric, filling bad rows with a sequential index
    df[scan_col] = pd.to_numeric(df[scan_col], errors='coerce')
    if df[scan_col].isna().any():
        df[scan_col] = df[scan_col].fillna(pd.Series(np.arange(len(df))))
    
    # 2. Identify the sensor column
    sensor_cols = [col for col in df.columns if any(k in col.lower() for k in ['°c', '101', 'ch_00', 'temp', '(c)'])]
    
    # Fallback
    if not sensor_cols:
        exclude = [scan_col] + [col for col in df.columns if 'time' in col.lower() or 'unnamed' in col.lower()]
        possible_sensors = [col for col in df.columns if col not in exclude]
        if possible_sensors:
            sensor_cols = [possible_sensors[0]]
        else:
            sensor_cols = [df.columns[-1]] # Absolute last resort
            
    sensor_col = sensor_cols[0]
    df[sensor_col] = pd.to_numeric(df[sensor_col], errors='coerce')
        
    return df, sensor_col, scan_col

# ====================================================================
# 2. STATE-AWARE PHYSICS ENGINE
# ====================================================================
def apply_physics_model(df: pd.DataFrame, sensor_col: str, dt_seconds: float = 1.0) -> pd.DataFrame:
    """
    Evaluates each sample against thermodynamic amplitude and rate limits[cite: 40].
    Strictly annotates data without filtering[cite: 29]. Vetoed samples do not update baseline state[cite: 42].
    """
    # Physics Configuration
    TAU = 15.0               # Thermal time constant for sheathed probe[cite: 34]
    TAU_MARGIN = 1.2         # Margin for tau uncertainty
    T_DRIVE_MAX = 2200.0     # Adiabatic flame temperature[cite: 40]
    T_SINK = 20.0            # Coldest available sink[cite: 40]
    
    processed_df = df.copy()
    processed_df['Quality_Flag'] = 'VALID'
    processed_df['dT_dt'] = np.nan
    
    # State tracking variables
    last_valid_T: Optional[float] = None
    
    for idx, row in processed_df.iterrows():
        T_raw = row[sensor_col]
        
        if pd.isna(T_raw):
            continue
            
        is_vetoed = False
        flags: List[str] = []
        
        # 1. Absolute Thermodynamic Amplitude Bounds[cite: 40]
        if T_raw > T_DRIVE_MAX or T_raw < -50.0:
            is_vetoed = True
            flags.append("Amplitude Out of Bounds")
            
        # 2. First-Order Lag Rate Limits[cite: 40]
        if not is_vetoed and last_valid_T is not None:
            dT_dt = (T_raw - last_valid_T) / dt_seconds
            processed_df.at[idx, 'dT_dt'] = dT_dt
            
            # Calculate absolute physics speed limits for this specific temperature
            max_slope = ((T_DRIVE_MAX - last_valid_T) / TAU) * TAU_MARGIN
            min_slope = (-(last_valid_T - T_SINK) / TAU) * TAU_MARGIN
            
            if dT_dt > max_slope or dT_dt < min_slope:
                is_vetoed = True
                flags.append(f"Rate Violation ({dT_dt:.1f} K/s)")

        # 3. State Update & Annotation[cite: 29, 42]
        if is_vetoed:
            processed_df.at[idx, 'Quality_Flag'] = "IMPLAUSIBLE: " + " | ".join(flags)
        else:
            last_valid_T = T_raw
            
    return processed_df

# ====================================================================
# 3. VISUALIZATION ENGINE
# ====================================================================
def plot_raw_data(df: pd.DataFrame, sensor_col: str, time_col: str, file_name: str):
    """Generates a simple line plot of the raw data before validation."""
    fig = go.Figure()
    fig.add_trace(go.Scatter(
        x=df[time_col], 
        y=df[sensor_col], 
        mode='lines', 
        name='Raw Temperature', 
        line=dict(color='#5454F5', width=1.5)
    ))
    
    fig.update_layout(
        title=f"Raw Sensor Data: {file_name}",
        xaxis_title="Scan Number",
        yaxis_title="Temperature (°C)",
        template='plotly_white',
        hovermode="x unified"
    )
    fig.show()

def plot_physics_results(df: pd.DataFrame, sensor_col: str, time_col: str, file_name: str):
    """Generates an interactive Plotly overlay of valid vs implausible data."""
    anomalies = df[df['Quality_Flag'] != 'VALID']
    valid_data = df[df['Quality_Flag'] == 'VALID']
    
    fig = go.Figure()
    
    # 1. Base Signal Line
    fig.add_trace(go.Scatter(
        x=df[time_col], 
        y=df[sensor_col], 
        mode='lines', 
        name='Raw Trajectory', 
        line=dict(color='lightgrey', width=1.5),
        hoverinfo='skip'
    ))
    
    # 2. Valid Physical Data
    fig.add_trace(go.Scatter(
        x=valid_data[time_col], 
        y=valid_data[sensor_col], 
        mode='markers', 
        name='Valid (Obeys Physics)', 
        marker=dict(color='#1f77b4', size=4)
    ))
    
    # 3. Implausible Anomalies (Vetoed)
    if not anomalies.empty:
        fig.add_trace(go.Scatter(
            x=anomalies[time_col], 
            y=anomalies[sensor_col], 
            mode='markers', 
            name='IMPLAUSIBLE (Physics Violation)', 
            marker=dict(color='red', size=8, symbol='x', line=dict(width=1.5, color='darkred')),
            text=anomalies['Quality_Flag'],
            hovertemplate="Scan: %{x}<br>Temp: %{y}°C<br>Flag: %{text}<extra></extra>"
        ))
        
    fig.update_layout(
        title=f"Physics-Based Anomaly Validation: {file_name}",
        xaxis_title="Scan Number",
        yaxis_title="Temperature (°C)",
        template='plotly_white',
        hovermode="x unified",
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
    )
    
    fig.show()

# ====================================================================
# 4. EXECUTION
# ====================================================================
file_path = r"C:\Users\hari7\Documents\Anamoly Detection\Anomaly-Dashboard\data\cataluminiscence\20260613T095926_21-11-2025 0 2025-11-21_cleaned.csv"
if os.path.exists(file_path):
    print(f"Loading {os.path.basename(file_path)}...")
    
    # 1. Extract
    df_raw, target_sensor, scan_col = load_and_parse_csv(file_path)
    print(f"Successfully loaded. Processing {len(df_raw)} samples on channel '{target_sensor}'.")
    
    # 2. Visualize Raw Data
    plot_raw_data(df_raw, target_sensor, scan_col, os.path.basename(file_path))
    
    # 3. Process Physics Validation (Assuming roughly 1 second per scan for dT/dt)
    df_validated = apply_physics_model(df_raw, target_sensor, dt_seconds=1.0)
    
    anomalies_count = (df_validated['Quality_Flag'] != 'VALID').sum()
    print(f"Physics Engine complete. Flagged {anomalies_count} implausible anomalies.")
    
    # 4. Visualize Validated Data
    plot_physics_results(df_validated, target_sensor, scan_col, os.path.basename(file_path))
else:
    print(f"File not found: {file_path}")
```
